Calculate the mean of values: 1, 2, 5




```python
values = [1, 2, 5]
mean_value = sum(values) / len(values)
print(f"The mean of {values} is: {mean_value}")
```

Calculate the magnitude of vector (1, 2, 5)

Magnitude = $\sqrt{1^2 + 2^2 + 5^2}$

```python
(1**2 + 2**2 + 5**2)** (1/2)
```

Normalize the vector

*Divide each component by the magnitude (and save to a new variable)*

```python
import math

x = 1
y = 2
z = 5
magnitude = math.sqrt(x**2 + y**2 + z**2)

normalized_x = x / magnitude
normalized_y = y / magnitude
normalized_z = z / magnitude

print(f"Normalized vector: ({normalized_x:.4f}, {normalized_y:.4f}, {normalized_z:.4f})")
```

Calculate the magnitude of the normalized vector

```python
import math
normalized_magnitude = math.sqrt(normalized_x**2 + normalized_y**2 + normalized_z**2)
print(f"The magnitude of the normalized vector is: {normalized_magnitude:.4f}")
```
